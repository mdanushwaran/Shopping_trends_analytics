```python

plt.figure(figsize=(6,6))
sns.countplot(data=data,x='Gender',order=data['Gender'].value_counts().index)
plt.title('Customer Count of Gender')
plt.xlabel('Gender', color = 'Brown',fontsize = 14, fontweight = 'bold')
plt.ylabel('Number of Customer',color = 'Brown',fontsize = 14, fontweight = 'bold')
plt.tick_params(axis = 'both' , labelsize = 10 , labelrotation = 0 , labelcolor = "lightgreen")
plt.subplots_adjust(left=0.2, bottom=0.2)
plt.show()
```

```python
## Pie chart of Payment Method.

payment_method = data['Payment Method'].value_counts()

plt.figure(figsize=(5, 5))
plt.title("Payment Method Distribution",
          color= 'Brown',
          fontsize = 14 ,
          family="Arial",
          fontweight = "bold")
plt.pie(payment_method, labels=payment_method.index,autopct='%1.1f%%',
        shadow = True , startangle=100)
plt.show()
```

```python
##  Line chart of Age vs Customer Count.

Age_count = data['Age'].value_counts()

plt.figure(figsize=(9,9))
plotprops = {
    'color': 'brown',
    'fontsize': 14,
    'family': 'Arial',
    'fontweight': 'bold',

}
plt.title("Customer Age Distribution",** plotprops)
plt.xlabel("Age", ** plotprops)
plt.ylabel("Count of Age distribution", **plotprops)
plt.plot(Age_count, label = 'Age_count', linestyle = 'dotted',
         marker = "o",
         markerfacecolor = 'black',
         markersize = 8,
         markeredgecolor = 'Red',
         color = 'blue')
plt.show()
```

```python
## Histogram of Age.

Age_count
```

```python
Age_count = data['Age'].value_counts()

plt.figure(figsize=(10,6))
plt.hist(Age_count , bins = 50)
plt.title("Age Distribution",** plotprops)
plt.xlabel("Age", ** plotprops)
plt.ylabel("series of Age distribution", **plotprops)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
## Count plot of Category.

Category = data['Category'].value_counts()

plt.figure(figsize=(6,6))
plt.bar(Category.index,Category.values,color = 'blue',edgecolor = 'black')
plt.title("Customer Count by Category",** plotprops)
plt.xlabel("Category", ** plotprops)
plt.ylabel("Count by Category", ** plotprops)
plt.tick_params(axis = 'both' , labelsize = 10 , labelrotation = 0 , labelcolor = "lightgreen")
plt.subplots_adjust(left=0.2, bottom=0.2)

for i, value in enumerate(Category.values):
    plt.text(i, value + 0.1, str(value), ha='center', fontsize=10)
plt.show()
```
