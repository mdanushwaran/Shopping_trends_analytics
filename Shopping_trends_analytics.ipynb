```python
import matplotlib.pyplot as plt

# Count each payment method
payment_count = data['Payment Method'].value_counts()

# Create pie chart
plt.figure(figsize=(4,4))
plt.pie(payment_count,
        labels=payment_count.index,
        autopct='%1.1f%%',
        startangle=100)

plt.title('Payment Method Distribution')
plt.show()
```

```python
data.groupby('Size')['Customer ID'].count()
```

```python
Size_count = data['Size'].value_counts()

plt.figure(figsize=(4,4))
plt.pie(Size_count , labels=Size_count.index , autopct = '%1.01f%%')
plt.title('Size Method')
plt.show()
```
